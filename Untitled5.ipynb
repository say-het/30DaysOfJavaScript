````python
import torch
import json
import os
import gc
from diffusers import FluxPipeline, LTXImageToVideoPipeline
from diffusers.utils import load_image, export_to_video
from PIL import Image
# Updated moviepy v2.0+ imports
from moviepy import VideoFileClip, AudioFileClip, concatenate_videoclips, CompositeAudioClip
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import numpy as np
import scipy.io.wavfile as wavfile

# Check GPU
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"VRAM: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
else:
    print("CUDA is not available. Please ensure you are running on a GPU runtime.")
````

````python
# ============================================================
# CELL 3 — Configuration
# ============================================================
CONFIG = {
    "paper_text": """
    # PASTE YOUR RESEARCH PAPER / LECTURE NOTES / REPORT TEXT HERE

    Abstract: This paper investigates the impact of climate change on global food security.
    We analyze data from 150 countries over 30 years and find that rising temperatures
    correlate with a 23% reduction in crop yields. Our findings suggest immediate policy
    intervention is required. Key findings include increased drought frequency, soil
    degradation, and shifting agricultural zones. We recommend a three-pronged approach:
    adaptation, mitigation, and international cooperation.
    """,

    "num_scenes": 4,          # How many scenes to generate
    "video_height": 480,
    "video_width": 704,
    "num_frames": 73,         # ~3 seconds at 24fps (must be 8n+1: 25,49,73,97)
    "fps": 24,
    "output_dir": "/content/hackathon_output"
}

os.makedirs(CONFIG["output_dir"], exist_ok=True)
os.makedirs(f"{CONFIG['output_dir']}/scenes", exist_ok=True)
print("Config ready ✅")
````

````python
# ============================================================
# CELL 4 — Step 1: Generate Script using Mistral
# ============================================================
def generate_script(paper_text, num_scenes):
    if not torch.cuda.is_available():
        print("❌ ERROR: GPU NOT FOUND.")
        print("Please go to Runtime -> Change runtime type -> Hardware accelerator and select 'GPU'.")
        print("Falling back to manual script...")
        return fallback_scenes(num_scenes)

    print("Loading Mistral for script generation...")

    tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-Instruct-v0.3")
    model = AutoModelForCausalLM.from_pretrained(
        "mistralai/Mistral-7B-Instruct-v0.3",
        torch_dtype=torch.float16,
        device_map="auto"
    )

    prompt = f"""[INST] You are a video script writer. Given this text, create a {num_scenes}-scene video script.

Return ONLY a valid JSON array like this:
[
  {{
    "scene_number": 1,
    "title": "Introduction",
    "narration": "Narration text here for text to speech. Keep it under 30 words.",
    "visual_prompt": "Detailed image generation prompt describing what should be shown visually. Be specific about style, colors, layout.",
    "duration_seconds": 3,
    "source_section": "Abstract"
  }}
]

Text to convert:
{paper_text}

Return only the JSON array, nothing else. [/INST]"""

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1500,
            temperature=0.7,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    response = tokenizer.decode(outputs[0][inputs['input_ids'].shape[1]:], skip_special_tokens=True)

    # Free VRAM
    del model, tokenizer
    torch.cuda.empty_cache()
    gc.collect()

    # Parse JSON
    try:
        clean_response = response.strip()
        if clean_response.startswith('```json') and clean_response.endswith('```'):
            clean_response = clean_response[7:-3].strip()
        elif clean_response.startswith('```') and clean_response.endswith('```'):
            clean_response = clean_response[3:-3].strip()

        start = clean_response.find('[')
        end = clean_response.rfind(']') + 1

        if start == -1 or end == -1 or start >= end:
            raise ValueError("Could not find a valid JSON array in the response.")

        json_str = clean_response[start:end]
        scenes = json.loads(json_str)
        print(f"✅ Script generated: {len(scenes)} scenes")
        return scenes
    except Exception as e:
        print(f"JSON parse failed: {e}")
        return fallback_scenes(num_scenes)

def fallback_scenes(num_scenes):
    """Fallback if LLM fails or GPU is missing"""
    scenes = []
    titles = ["Introduction", "Key Findings", "Methodology", "Conclusion"]
    narrations = [
        "Welcome to this research explainer. Today we explore an important scientific study.",
        "The key findings reveal significant patterns in the data collected over thirty years.",
        "Researchers analyzed data from one hundred fifty countries using advanced statistical methods.",
        "In conclusion, immediate action is required based on these compelling findings."
    ]
    visuals = [
        "A clean academic title slide with globe icon, white background, professional blue typography",
        "A bar chart showing data trends over 30 years, minimalist infographic style, blue palette",
        "A world map with data points highlighted across 150 countries, research visualization style",
        "A summary infographic with three key action items, clean design, green and blue colors"
    ]
    for i in range(min(num_scenes, 4)):
        scenes.append({
            "scene_number": i+1,
            "title": titles[i],
            "narration": narrations[i],
            "visual_prompt": visuals[i],
            "duration_seconds": 3,
            "source_section": f"Section {i+1}"
        })
    return scenes

# RUN IT
scenes = generate_script(CONFIG["paper_text"], CONFIG["num_scenes"])

with open(f"{CONFIG['output_dir']}/script.json", "w") as f:
    json.dump(scenes, f, indent=2)

print("\n📄 GENERATED SCRIPT:")
for scene in scenes:
    print(f"\nScene {scene['scene_number']}: {scene['title']}")
    print(f"  Narration: {scene['narration']}")
````

````python
# ============================================================
# CELL 6 — Step 2: Generate Images
# ============================================================
from diffusers import AutoPipelineForText2Image

def generate_scene_images(scenes, regenerate_scene=None):
    print("Loading Image Generation Pipeline...")

    # Using SDXL Turbo as a high-speed, non-gated alternative to FLUX
    pipe = AutoPipelineForText2Image.from_pretrained(
        "stabilityai/sdxl-turbo",
        torch_dtype=torch.float16,
        variant="fp16"
    )
    pipe.to("cuda")

    for scene in scenes:
        scene_num = scene["scene_number"]
        output_path = f"{CONFIG['output_dir']}/scenes/scene_{scene_num}_image.png"

        if regenerate_scene is not None and scene_num != regenerate_scene:
            continue

        if regenerate_scene is None and os.path.exists(output_path):
            print(f"  Scene {scene_num} image exists, skipping...")
            continue

        print(f"  Generating image for Scene {scene_num}...")

        # SDXL Turbo works best with 1-4 steps and guidance_scale 0.0 or 1.0
        image = pipe(
            prompt=scene["visual_prompt"],
            num_inference_steps=2,
            guidance_scale=0.0,
            height=CONFIG["video_height"],
            width=CONFIG["video_width"],
        ).images[0]

        image.save(output_path)
        print(f"  ✅ Saved: {output_path}")

    del pipe
    torch.cuda.empty_cache()
    gc.collect()
    print("✅ All images generated")

# RUN IT
generate_scene_images(scenes)
````

````python
# ============================================================
# CELL 7 — Step 3: Generate Narration Audio with Kokoro
# ============================================================
def generate_narration(scenes, regenerate_scene=None):
    print("Loading Kokoro TTS...")

    try:
        from kokoro import KPipeline
        tts = KPipeline(lang_code='a')  # 'a' = American English
        use_kokoro = True
    except:
        print("Kokoro not available, falling back to gTTS...")
        use_kokoro = False

    for scene in scenes:
        scene_num = scene["scene_number"]
        output_path = f"{CONFIG['output_dir']}/scenes/scene_{scene_num}_audio.wav"

        if regenerate_scene is not None and scene_num != regenerate_scene:
            continue

        if regenerate_scene is None and os.path.exists(output_path):
            print(f"  Scene {scene_num} audio exists, skipping...")
            continue

        print(f"  Generating audio for Scene {scene_num}...")

        if use_kokoro:
            generator = tts(
                scene["narration"],
                voice='af_heart',  # warm female voice
                speed=1.0
            )

            audio_chunks = []
            for i, (gs, ps, audio) in enumerate(generator):
                audio_chunks.append(audio)

            if audio_chunks:
                full_audio = np.concatenate(audio_chunks)
                wavfile.write(output_path, 24000, full_audio)
        else:
            # Fallback: gTTS
            from gtts import gTTS
            tts_fallback = gTTS(text=scene["narration"], lang='en')
            mp3_path = output_path.replace('.wav', '.mp3')
            tts_fallback.save(mp3_path)
            # Convert mp3 to wav using moviepy
            AudioFileClip(mp3_path).write_audiofile(output_path)
            os.remove(mp3_path)

        print(f"  ✅ Saved: {output_path}")

    print("✅ All narration generated")

# RUN IT
generate_narration(scenes)
````

````python
# ============================================================
# CELL 8 — Step 4: Generate Video Clips with LTX-Video
# ============================================================
def generate_scene_videos(scenes, regenerate_scene=None):
    print("Loading LTX-Video...")

    ltx_pipe = LTXImageToVideoPipeline.from_pretrained(
        "Lightricks/LTX-Video",
        torch_dtype=torch.bfloat16
    )
    ltx_pipe.enable_model_cpu_offload()

    for scene in scenes:
        scene_num = scene["scene_number"]
        image_path = f"{CONFIG['output_dir']}/scenes/scene_{scene_num}_image.png"
        output_path = f"{CONFIG['output_dir']}/scenes/scene_{scene_num}_video.mp4"

        if regenerate_scene is not None and scene_num != regenerate_scene:
            continue

        if regenerate_scene is None and os.path.exists(output_path):
            print(f"  Scene {scene_num} video exists, skipping...")
            continue

        if not os.path.exists(image_path):
            print(f"  ☑ᄅ No image found for scene {scene_num}, skipping...")
            continue

        print(f"  Animating Scene {scene_num}: {scene['title']}...")

        keyframe = load_image(image_path)

        frames = ltx_pipe(
            image=keyframe,
            prompt=f"Smooth professional animation. {scene['visual_prompt']}. Slow gentle camera motion.",
            negative_prompt="shaky, blurry, distorted, low quality, flickering",
            width=CONFIG["video_width"],
            height=CONFIG["video_height"],
            num_frames=CONFIG["num_frames"],
            num_inference_steps=50,
            guidance_scale=3.0,
        ).frames[0]

        export_to_video(frames, output_path, fps=CONFIG["fps"])
        print(f"  ✅ Saved: {output_path}")

    del ltx_pipe
    torch.cuda.empty_cache()
    gc.collect()
    print("✅ All video clips generated")

# RUN IT - Added logic to load scenes if variable is missing
try:
    generate_scene_videos(scenes)
except NameError:
    with open(f"{CONFIG['output_dir']}/script.json", "r") as f:
        scenes = json.load(f)
    generate_scene_videos(scenes)
````

````python
# ============================================================
# CELL 9 — Step 5: Stitch Everything into Final Video
# ============================================================
def stitch_final_video(scenes, output_name="final_video.mp4"):
    print("Stitching final video...")

    clips = []

    for scene in scenes:
        scene_num = scene["scene_number"]
        video_path = f"{CONFIG['output_dir']}/scenes/scene_{scene_num}_video.mp4"
        audio_path = f"{CONFIG['output_dir']}/scenes/scene_{scene_num}_audio.wav"

        if not os.path.exists(video_path):
            print(f"  ☑ᄅ Missing video for scene {scene_num}, skipping...")
            continue

        video_clip = VideoFileClip(video_path)

        # Attach audio if it exists
        if os.path.exists(audio_path):
            audio_clip = AudioFileClip(audio_path)

            # Match durations
            if audio_clip.duration > video_clip.duration:
                video_clip = video_clip.loop(duration=audio_clip.duration)
            else:
                video_clip = video_clip.subclip(0, audio_clip.duration)

            video_clip = video_clip.set_audio(audio_clip)

        clips.append(video_clip)
        print(f"  ✅ Scene {scene_num} added ({video_clip.duration:.1f}s)")

    if not clips:
        print("❌ No clips found!")
        return

    final = concatenate_videoclips(clips, method="compose")
    output_path = f"{CONFIG['output_dir']}/{output_name}"
    final.write_videofile(output_path, fps=CONFIG["fps"], codec="libx264", audio_codec="aac")

    for clip in clips:
        clip.close()
    final.close()

    print(f"\nᄇ FINAL VIDEO SAVED: {output_path}")
    return output_path

# RUN IT - Added logic to load scenes if variable is missing
try:
    final_path = stitch_final_video(scenes)
except NameError:
    with open(f"{CONFIG['output_dir']}/script.json", "r") as f:
        scenes = json.load(f)
    final_path = stitch_final_video(scenes)
````

````python
# ============================================================
# CELL 10 — 🔁 MODULAR REGENERATION (The Key Hackathon Feature)
# ============================================================
# This is what makes your system "agentic" —
# User edits ONE scene, only that scene reruns

SCENE_TO_EDIT = 2  # Change this to whichever scene number you want to edit

# Load current script
with open(f"{CONFIG['output_dir']}/script.json", "r") as f:
    scenes = json.load(f)

# ---- MAKE YOUR EDITS HERE ----
scene_idx = SCENE_TO_EDIT - 1  # 0-indexed

# Edit narration
scenes[scene_idx]["narration"] = "YOUR NEW NARRATION FOR THIS SCENE HERE"

# Edit visual
scenes[scene_idx]["visual_prompt"] = "YOUR NEW VISUAL DESCRIPTION HERE"

# ---- END EDITS ----

# Save updated script
with open(f"{CONFIG['output_dir']}/script.json", "w") as f:
    json.dump(scenes, f, indent=2)

print(f"✅ Scene {SCENE_TO_EDIT} updated. Regenerating only this scene...")

# Regenerate ONLY the edited scene
generate_scene_images(scenes, regenerate_scene=SCENE_TO_EDIT)
generate_narration(scenes, regenerate_scene=SCENE_TO_EDIT)
generate_scene_videos(scenes, regenerate_scene=SCENE_TO_EDIT)

# Re-stitch final video
final_path = stitch_final_video(scenes, output_name="final_video_v2.mp4")
print(f"\n✅ Done! Only Scene {SCENE_TO_EDIT} was regenerated.")
````
